```python
import os
from glob import glob
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
reallog = sorted(glob('real/*csv'),key=lambda x: int(x.split('/')[1].split('_')[0]))
fakelog = "fake/ol3_rl1/model_log.csv"
stateveclog = "statevector/model_log.csv"
```

```python
# Statevector
svdf = pd.read_csv(stateveclog,delimiter=',').dropna()[['Iteration','Cost']]

# Fake
fakedf = pd.read_csv(fakelog,delimiter=',').dropna()[['Iteration','Cost']]

# Real
realdf = pd.concat([pd.read_csv(i,delimiter=',').dropna()['Cost'] for i in reallog]).reset_index(drop=True).reset_index().rename(columns={"index":'Iteration'})
realdf['Iteration'] += 1
```

```python
cmap = sns.color_palette('Paired',6)
```

```python
cmap
```

```python
sns.lineplot(svdf,x='Iteration',y='Cost',label='State-vector',color=cmap[5])
sns.lineplot(fakedf,x='Iteration',y='Cost',label=r'$\it{FakeQuebec}$',color=cmap[3])
sns.lineplot(realdf,x='Iteration',y='Cost',label=r'$\it{ibm\_quebec}$',color=cmap[1])


# plt.hlines(svdf['Cost'].iloc[-1],-5,506,color=cmap[5])
# plt.hlines(fakedf['Cost'].iloc[-1],-5,506,color=cmap[3])
# plt.hlines(realdf['Cost'].iloc[-1],-5,506,color=cmap[1])
plt.ylim(0)
plt.xlim(-4,505)
plt.xlabel("Iterations")
plt.ylabel("Loss (unitless)")
plt.tight_layout()
plt.savefig(os.path.join(os.path.expanduser('~'),'qregress/images/DDCC/ibmq_vs_statevector_vs_fake_Loss.png'),dpi=300,bbox_inches='tight')
plt.show()
```
